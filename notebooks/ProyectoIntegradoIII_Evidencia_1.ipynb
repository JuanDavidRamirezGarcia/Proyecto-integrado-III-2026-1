## I. Definición del problema de negocio

**Definición de problema y su impacto**

Según la **FAO** los precios mundiales de productos alimenticios subieron en Marzo del 2026 por segundo mes consecutivo. Debido en gran medidad por el conflicto que hay en medio oriente, el cual ha encarecido los precios enérgeticos y por ende el precio de alimentos, lo cuales cuestan más su producción. Como muestra el índice de precios subió un **2.4%** más comparado con febrero del mismo año y **1%** comparado con el promedio del año anterior. Se espera que los precios sigan subiendo a medida que el conflicto continue.

Para Latinoamerica el impacto será el mismo, el alza de precios de productos de la canasta básica, ya que la región depende de fertilizantes importados para la producción, lo cuál aumento el costo de producción y cultivo. En consecuencias estos precios se trasladan al consumidor impacto el alza del costo de vida.

De acá la importancia de analizar los precios de algunos alimentos en Colombia y compararlos con los precios de otros paises en la región.

Fuente: https://www.fao.org/newsroom/detail/fao-food-price-index-rises-in-march-as-near-east-conflict-raises-energy-costs/es



**Pregunta de investigación:**

¿Cómo varió el costo de la "Canasta de Desayuno" (Breakfast Basket) en las ciudades de Colombia comparado con el promedio regional de Latinoamérica entre octubre 2025 y marzo 2026, y qué categorías de alimentos fueron las más volátiles?

**Definición de métricas para evaluar el éxito del análisis.**

1. Puntaje de encarecimiento: Esta métrica mide el inidice de encarecemiento acumulado. Asiganandoles a todos los paises un puntaje base de 100 puntos y compara los precios a través del tiempo.

$Puntaje=(\frac{Price-USD-actual}{Price-USD-octubre-2025})*100$

2. Variación del costo de la canasta familiar: Esta métrica nos mostrara el porcentaje total de encarecmiento de los productos de la canasta básica.

$variacion=(\frac{Breakfast-Basket-Marzo2026-Breakfast-Basket-Octubre2025}{Breakfast-Basket-Octubre2025})*100$

3. Diferencia del costo de la canasta en dolares= En esta variable se va a comparar el costo en dolares de la canasta básica con el promedio de los paise de Latinoamérica.

$\text{Brecha (USD)} = \text{Breakfast-Basket-USD}_{Colombia} - \text{Promedio-Basket-USD}_{\text{Latam}}$

## II Fuente de datos

Los datos fueron encontrados en Kaggle. Nombre del dataset: Global **Grocery Inflation (2025/2026)**

Enlace del data set: https://www.kaggle.com/datasets/waddahali/global-grocery-inflation-20252026/data

**Descripción del conjunto de datos:**

* City: Nombre de la ciudad.

* Country: Nombre del país.

* ISO_Country_Code: Código internacional de 3 letras del país.

* Region: Subdivisión geográfica o región del país.

* Continent: Continente al que pertenece el país.

* Month: Fecha del registro por año y mes.

* Month_Name: Nombre completo del mes y año.

* Item: Nombre del alimento específico.

* Item_Key: ID o identificador del producto.

* Item_Category: Categoría del producto si es Lácteos, Carnes, etc..

* Price_USD: Precio del producto normalizado a dólares.

* Unit: Unidad de medida si es kg, litro, unidad.

* Currency: Moneda original antes de la conversión a dolares.

* Exchange_Rate: Valor usado para convertir la moneda local a USD.

* Inflation_Source: Fuente de los datos macroeconómicos de referencia.

* FAO_Index_Value: Valor del índice global de precios de la FAO.

* FAO_Index_Date: Fecha de publicación del índice FAO.

* FAO_YoY_Change_Pct: Variación porcentual anual del índice global.

* USDA_All_Food_Forecast_Pct: Inflación de alimentos proyectada por EE.UU.

* USDA_Food_At_Home_Pct: Inflación proyectada para comida comprada para el hogar.

* Data_Collection_Date: Día exacto en que se tomó el precio en la tienda.

* Source_URL: Enlace web a la fuente del precio.

* Population_Estimate: Cantidad de habitantes estimada de la ciudad.

* Breakfast_Basket_USD: Costo total de una canasta básica en la ciudad.

* Latitud: Coordenada de ubicación norte-sur.

* Longitud: Coordenada de ubicación este-oeste.

* Is_Capital: Indica con SI/NO si la ciudad es la capital del país.

   * `Breakfast_Basket_USD`: podemos obervar que la mayoría de los gastos en la canasta básica se concentran en un rango moderado de 5 a 10 dólares, mientras que los extremos altos entre 15 y 20 dólares son menos frecuentes y reflejan consumos de aliemntos caros.
